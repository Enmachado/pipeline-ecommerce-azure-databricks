```python
from pyspark.sql.functions import col, to_date, upper

# 1. Configuração de acesso (mesma do 01_bronze)
spark.conf.set(
    "SUA_CHAVE_AQUI"
)

# 2. Ler o CSV (Bronze)
df_bronze = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("SUA_CHAVE_AQUI/vendas/vendas_ecommerce.csv")

# 3. Transformações Silver
df_silver = df_bronze \
    .dropna() \
    .filter(col("status_pedido") != "Cancelado") \
    .withColumn("data_pedido", to_date(col("data_pedido"), "yyyy-MM-dd")) \
    .withColumn("receita_liquida", col("valor_total") + col("frete")) \
    .withColumn("estado", upper(col("estado"))) \
    .withColumn("quantidade", col("quantidade").cast("int")) \
    .withColumn("valor_total", col("valor_total").cast("double")) \
    .withColumn("frete", col("frete").cast("double"))


df_silver.write.format("delta").mode("overwrite") \
    .save("SUA_CHAVE_AQUI/vendas/silver/vendas")


print(f"Total de registros Silver: {df_silver.count()}")
display(df_silver)
```
